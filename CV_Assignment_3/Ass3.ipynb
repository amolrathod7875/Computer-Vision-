```python
import cv2
import matplotlib.pyplot as plt

def process_image(image_path):
    # Step 3: Load the image (OpenCV loads in BGR format by default)
    image_bgr = cv2.imread(image_path)

    # Step 4: Check if image is loaded
    if image_bgr is None:
        print("Error: Could not load the image. Please check the file path.")
        return

    # Step 5: Convert BGR to RGB
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    # Step 6: Convert RGB to Grayscale
    image_gray = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

    # Step 7: Convert RGB to HSV
    image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

    # Step 8: Split HSV channels to show their significance
    h_channel, s_channel, v_channel = cv2.split(image_hsv)

    # Step 9: Display the results
    plt.figure(figsize=(12, 8))

    # Original RGB
    plt.subplot(2, 3, 1)
    plt.imshow(image_rgb)
    plt.title('Original RGB')
    plt.axis('off')

    # Grayscale
    plt.subplot(2, 3, 2)
    plt.imshow(image_gray, cmap='gray')
    plt.title('Grayscale (Intensity)')
    plt.axis('off')

    # Full HSV (Visualized as RGB for structural differences)
    plt.subplot(2, 3, 3)
    plt.imshow(image_hsv)
    plt.title('HSV Color Space')
    plt.axis('off')

    # Hue Channel
    plt.subplot(2, 3, 4)
    plt.imshow(h_channel, cmap='gray')
    plt.title('Hue Channel (Color Type)')
    plt.axis('off')

    # Saturation Channel
    plt.subplot(2, 3, 5)
    plt.imshow(s_channel, cmap='gray')
    plt.title('Saturation Channel (Purity)')
    plt.axis('off')

    # Value Channel
    plt.subplot(2, 3, 6)
    plt.imshow(v_channel, cmap='gray')
    plt.title('Value Channel (Brightness)')
    plt.axis('off')

    plt.tight_layout()
    plt.show()

# Run the function (Replace 'your_image.jpg' with an actual image file)
process_image('BMW.jpeg')
```
